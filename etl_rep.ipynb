```python
import pandas as pd

# Raw US House election data
raw_df = pd.read_csv('data/rep.csv', sep='\t', header=None, names=['raw_text'])

# Separate data into columns and filter for 2024 results only
clean_df = raw_df['raw_text'].str.split(',', expand=True)
header = clean_df.iloc[0]

data_rows = clean_df.iloc[1:]
df_2024 = data_rows[data_rows[0] == '2024']

# Clean data length
df_2024 = df_2024.iloc[:, :len(header)]
df_2024.columns = header
df_2024 = df_2024.iloc[:, :-2]

df_2024.to_csv('data/rep_2024.csv', index=False)
df_2024.head()
```

```python
df = pd.read_csv('data/rep_2024.csv')

# Reset index
df.reset_index(drop=True, inplace=True)

# Convert column B to string, pad it to 2 digits, and combine with column A
df['district'] = df['state_po'] + df['district'].astype(str).str.zfill(2)

# Drop unnecessary data
columns_to_drop = ['state_fips', 'state_cen', 'state_ic', 'state_po', 'office', 'writein', 'mode', 'unofficial', 'version', 'fusion_ticket']
df = df.drop(columns=columns_to_drop)

# Capitalize columns
df.columns = df.columns.str.capitalize()

df_2024.to_csv('data/rep_2024.csv', index=False)
df.head()
```
