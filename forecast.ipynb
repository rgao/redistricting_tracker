```python
import numpy as np
import pandas as pd

# 2026 Simulation Data: Raw polls for two highly competitive House districts
raw_polls_data = {
    'district': ['PA-07', 'PA-07', 'MI-07', 'MI-07'],
    'pollster': ['Pollster_Alpha', 'Pollster_Beta', 'Pollster_Alpha', 'Pollster_Gamma'],
    'dem_lead': [0.03, -0.01, 0.02, 0.05],  # Positive = Dem lead, Negative = Rep lead
    'sample_size': [1000, 800, 1200, 900],
    'days_old':,              # Days since the poll field work ended
    'pollster_rating': [0.9, 0.6, 0.9, 0.4]  # Internal quality score between 0 and 1
}
polls_df = pd.DataFrame(raw_polls_data)

# Constants for weight calculations
decay_constant = 0.05  # Controls how quickly old polls lose value

def calculate_poll_weights(df):
    # Variance weight component (proportional to the square root of sample size)
    df['weight_sample'] = np.sqrt(df['sample_size'])
    
    # Recency weight component (exponential decay function)
    df['weight_recency'] = np.exp(-decay_constant * df['days_old'])
    
    # Combined total weight
    df['total_weight'] = df['weight_sample'] * df['weight_recency'] * df['pollster_rating']
    return df

weighted_polls = calculate_poll_weights(polls_df)

# Calculate the final weighted polling mean for each district
aggregated_polls = weighted_polls.groupby('district').apply(
    lambda x: np.average(x['dem_lead'], weights=x['total_weight'])
).to_dict()

print("Aggregated Polling Blends (Dem Margin):", aggregated_polls)
```
